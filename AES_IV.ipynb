```python
!pip install pycryptodome
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
import os

def encrypt_message(message, key, filename="encrypted_message.bin"):
    cipher = AES.new(key, AES.MODE_CBC)
    ciphertext = cipher.encrypt(pad(message.encode(), AES.block_size))
    with open(filename, "wb") as f:
        f.write(cipher.iv)
        f.write(ciphertext)
    return ciphertext, cipher.iv

def decrypt_message(ciphertext, key, iv, filename="encrypted_message.bin"):
    with open(filename, "rb") as f:
        iv = f.read(AES.block_size)
        ciphertext = f.read()
    cipher = AES.new(key, AES.MODE_CBC, iv=iv)
    plaintext = unpad(cipher.decrypt(ciphertext), AES.block_size)
    return plaintext.decode()

if __name__ == "__main__":
    message = input("Enter a message to encrypt: ")
    key = os.urandom(32)
    ciphertext, iv = encrypt_message(message, key)
    decrypted_message = decrypt_message(ciphertext, key, iv)
    print("Encrypted message:", ciphertext.hex())
    print("Decrypted message:", decrypted_message)

```
