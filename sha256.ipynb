```python
import hashlib

password = input("Enter a password: ")

password = password.encode()

result = hashlib.sha256(password)

with open('hashed_passwords.txt','a') as file:
  file.write(result.hexdigest() + '\n')

print("Hashed password: " + result.hexdigest())
```
